## Raw data validation

This section checks whether all raw datasets can be accessed and
successfully read.

No cleaning, transformation, filtering, or scoring is performed here.
The purpose is only to validate data availability and file integrity.

```python
raw_files = sorted(
    file
    for file in cfg.RAW_DATA_DIR.iterdir()
    if file.is_file()
)

print(
    f"Files found in raw data: {len(raw_files)}"
)

for file in raw_files:
    print(f"- {file.name}")
```

```python
def read_raw_dataset(path: Path) -> pd.DataFrame:
    """
    Read a raw dataset according to its file extension.

    Supported formats:
        .sav
        .csv
        .xlsx

    No preprocessing or transformation is applied.
    """

    suffix = path.suffix.lower()

    if suffix == ".sav":
        return pd.read_spss(path)

    if suffix == ".csv":
        return pd.read_csv(path)

    if suffix == ".xlsx":
        return pd.read_excel(path)

    raise ValueError(
        f"Unsupported file format: {suffix}"
    )
```

```python
validation_results = []

raw_datasets = {}

for file in raw_files:

    try:

        df = read_raw_dataset(file)

        raw_datasets[file.name] = df

        validation_results.append(
            {
                "file": file.name,
                "format": file.suffix.lower(),
                "status": "OK",
                "rows": df.shape[0],
                "columns": df.shape[1],
                "missing_cells": int(
                    df.isna().sum().sum()
                ),
                "duplicated_rows": int(
                    df.duplicated().sum()
                ),
                "error": None,
            }
        )

    except Exception as error:

        validation_results.append(
            {
                "file": file.name,
                "format": file.suffix.lower(),
                "status": "ERROR",
                "rows": None,
                "columns": None,
                "missing_cells": None,
                "duplicated_rows": None,
                "error": str(error),
            }
        )


raw_validation = pd.DataFrame(
    validation_results
)

raw_validation
```

```python
n_files = len(raw_validation)

n_ok = (
    raw_validation["status"] == "OK"
).sum()

n_errors = (
    raw_validation["status"] == "ERROR"
).sum()


print(f"Raw files found:  {n_files}")
print(f"Successfully read: {n_ok}")
print(f"Read errors:       {n_errors}")
```

```python
if n_errors == 0:
    print(
        "\n✓ All raw datasets were read successfully."
    )
else:
    print(
        "\n⚠ Some raw datasets could not be read."
    )

    display(
        raw_validation.loc[
            raw_validation["status"] == "ERROR",
            ["file", "error"],
        ]
    )
```
